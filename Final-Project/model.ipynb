# Rreading Data from the train folder 

```python
import os 
import cv2 

from pathlib import Path
import pandas as pd

inputData = r'archive/seg_train'

imgDesc = []

for className in os.listdir(inputData):
    classPath = os.path.join(inputData, className)

    for imgName in os.listdir(classPath):
        imgPath = os.path.join(classPath, imgName)

        img = cv2.imread(imgPath)

        if img is not None:
            height, width, channels = img.shape
            imgType = img.dtype
            resolution = (height,width)

            imgDesc.append({
                'imgName' : imgName,
                'className' : className,
                'height' : height,
                'width' : width,
                'channels' : channels,
                'type' : imgType,
                'resolution' : resolution,
                'imgPath' : imgPath,
            })

imgDf = pd.DataFrame(imgDesc)
imgDf.to_csv('imgDesc.csv', index=False)
```
